```python
import math
from collections import Counter

def prob_draw(urn, target_draw):
    """
    urn: list of ball colours (e.g., ["white","white","black","red"])
    target_draw: dict specifying desired outcome 
                 (e.g., {"white": 1, "red": 1})
    """
    total_balls = len(urn)
    total_ways = math.comb(total_balls, sum(target_draw.values()))
    
    # count how many of each colour in urn
    urn_counts = Counter(urn)
    
    # favourable ways = product of combinations for each colour
    favourable = 1
    for colour, count_needed in target_draw.items():
        favourable *= math.comb(urn_counts[colour], count_needed)
    
    return Fraction(favourable / total_ways).limit_denominator()
```

```python
from collections import Counter
from math import comb
from fractions import Fraction
import random

# --- Core helpers ------------------------------------------------------------

def _as_counter(urn):
    """
    Accepts either:
      - dict like {"white": 4, "red": 3, "black": 5}
      - list like ["white","white",...]
    Returns Counter of colors.
    """
    if isinstance(urn, dict):
        return Counter(urn)
    return Counter(urn)  # list also works

def _draw_size(target_draw: dict) -> int:
    return sum(target_draw.values())

def prob_draw_from_urn(urn, target_draw) -> float:
    """
    Exact probability of drawing the target combo from a single urn (without replacement).
    urn: dict or list of colors
    target_draw: dict like {"white":1,"red":1}
    """
    counts = _as_counter(urn)
    k = _draw_size(target_draw)
    total = sum(counts.values())
    if k > total:
        return 0.0

    # total ways to choose k balls
    total_ways = comb(total, k)

    # favourable ways = product over colors of choose(count[color], need[color]),
    # provided we don't over-ask for any color
    fav = 1
    for color, need in target_draw.items():
        have = counts.get(color, 0)
        if need > have:
            return 0.0
        fav *= comb(have, need)

    return fav / total_ways

def posterior_by_urn(urns, target_draw, priors=None):
    """
    Bayes posterior P(urn_i | observed target_draw) for each urn_i.

    urns: list of urn specs (each dict or list)
    target_draw: dict like {"white":1,"red":1}
    priors: list of floats summing to 1; if None, assumes equal priors.

    Returns: dict {i: posterior_prob}, where i is urn index (0-based).
    """
    n = len(urns)
    if priors is None:
        priors = [1.0 / n] * n
    assert len(priors) == n and abs(sum(priors) - 1.0) < 1e-12

    likelihoods = [prob_draw_from_urn(u, target_draw) for u in urns]
    evidence = sum(p * L for p, L in zip(priors, likelihoods))
    if evidence == 0.0:
        # impossible observation under all urns
        return {i: 0.0 for i in range(n)}

    post = {i: (priors[i] * likelihoods[i]) / evidence for i in range(n)}
    return post

def posterior_for_subset(urns, target_draw, subset_indices, priors=None):
    """
    P(urn in subset | observed target_draw)
    subset_indices: iterable of urn indices (e.g., [1,2] for 'B or C')
    """
    post = posterior_by_urn(urns, target_draw, priors)
    return sum(post[i] for i in subset_indices)

# --- Nice-to-have: exact-as-Fraction and a simulator -------------------------

def prob_draw_from_urn_fraction(urn, target_draw) -> Fraction:
    """Same as prob_draw_from_urn, but returns a simplified Fraction."""
    counts = _as_counter(urn)
    k = _draw_size(target_draw)
    total = sum(counts.values())
    if k > total:
        return Fraction(0, 1)

    total_ways = comb(total, k)
    fav = 1
    for color, need in target_draw.items():
        have = counts.get(color, 0)
        if need > have:
            return Fraction(0, 1)
        fav *= comb(have, need)

    return Fraction(fav, total_ways)  # exact rational

def simulate_posterior(urns, target_draw, priors=None, trials=100_000, rng=None):
    """
    Monte Carlo estimate of P(urn_i | target) using random draws.
    """
    if rng is None:
        rng = random.Random()
    n = len(urns)
    if priors is None:
        priors = [1.0 / n] * n

    chosen_counts = [0] * n   # how often urn i produced the target
    target_hits = 0

    # Pre-expand any dict urns to lists for random sampling
    expanded = []
    for u in urns:
        if isinstance(u, dict):
            lst = []
            for c, k in u.items():
                lst.extend([c] * k)
            expanded.append(lst)
        else:
            expanded.append(list(u))

    k = _draw_size(target_draw)

    for _ in range(trials):
        # choose urn by priors
        r = rng.random()
        cum = 0.0
        i = 0
        for i, p in enumerate(priors):
            cum += p
            if r <= cum:
                break

        urn_list = expanded[i]
        if len(urn_list) < k:
            continue
        draw = rng.sample(urn_list, k)
        if Counter(draw) == Counter(target_draw):
            target_hits += 1
            chosen_counts[i] += 1

    if target_hits == 0:
        return {i: 0.0 for i in range(n)}
    return {i: chosen_counts[i] / target_hits for i in range(n)}

# --- Example usage -----------------------------------------------------------

if __name__ == "__main__":
    # You can specify urns as dicts (scales better than long lists):
    urn_a = {"white": 1, "black": 2, "red": 3}
    urn_b = {"white": 2, "black": 1, "red": 1}
    urn_c = {"white": 4, "black": 5, "red": 3}
    urns = [urn_a, urn_b, urn_c]

    # Target: 1 white and 1 red (two-ball draw)
    target = {"white": 1, "red": 1}

    # Posteriors with equal priors:
    post = posterior_by_urn(urns, target)
    print("Posterior by urn (equal priors):", post)

    # Probability it came from B or C (indices 1 and 2):
    p_b_or_c = posterior_for_subset(urns, target, subset_indices=[1, 2])
    print("P(B or C | target):", p_b_or_c)

    # Exact per-urn likelihoods as Fractions:
    print("Per-urn exact likelihoods:")
    for i, u in enumerate(urns):
        print(f"  Urn {i}: {prob_draw_from_urn_fraction(u, target)}")

    # Optional: Monte Carlo sanity check
    est = simulate_posterior(urns, target, trials=200_000)
    print("Monte Carlo posterior estimate:", est)
```
